```python
from utils.scoring import readability_score, keyword_match_score, length_score
import pandas as pd
import json
```

```python
# Load responses
with open('../responses/sample_responses.json') as f:
    data = json.load(f)

prompt = data["prompt"]
responses = data["responses"]
```

```python
# Evaluate each response
evaluations = []
for idx, response in enumerate(responses):
    score1 = readability_score(response)
    score2 = keyword_match_score(response, prompt)
    score3 = length_score(response)

    total_score = round(score1 + score2 + score3, 2)

    evaluations.append({
        "response_id": idx + 1,
        "response": response,
        "readability": score1,
        "keyword_score": score2,
        "length_score": score3,
        "total_score": total_score
    })
```

```python
# Convert to DataFrame
df = pd.DataFrame(evaluations)
df_sorted = df.sort_values(by='total_score', ascending=False)
```
